## Uma linguagem com tipagem dinâmica

Isso quer dizer que você não precisa declarar o tipo das variáveis. O tipo é inferido em tempo de execução.

```python
# Exemplo de tipagem dinâmica
variavel = 10  # Aqui, variável é um inteiro
print(f"O valor de 'variavel' é {variavel} e seu tipo é {type(variavel)}")

variavel = "Agora sou uma string!"  # Agora, variável é uma string
print(f"O valor de 'variavel' é '{variavel}' e seu tipo é {type(variavel)}")

```

# Mas com tipagem forte

Python não converte automaticamente tipos incompatíveis. Você precisa fazer conversões explícitas.

```python
# Exemplo de tipagem forte
numero = 42
texto = "O número é "

# Isso causaria um erro: print(texto + numero)
# Correção com conversão explícita
print(texto + str(numero))
```

# Programação imperativa
O paradigma imperativo é baseado em comandos que mudam o estado do programa. As variáveis `soma` e `item`, por exemplo mudam a medida que o progama é executado.



```python
# Exemplo de laço for
array = [1, 2, 3, 4, 5]
soma = 0
for item in array:
    if item % 2 == 0:
        soma = soma + item

print(f"A soma dos números pares é: {soma}")

```

#Programação funcional
Em Python, funções são [objetos de primeira classe](https://en.wikipedia.org/wiki/First-class_function): podem ser passadas como argumentos, retornadas por outras funções, modificadas e atribuídas a variáveis. A linguagem também suporta conceitos como funções puras e `lambdas`.

```python
# Usando list comprehension para somar números pares
array = [1, 2, 3, 4, 5]

soma_pares = sum([x for x in array if x % 2 == 0])
print(f"A soma dos números pares é: {soma_pares}")
```

Um exemplo de uso de lambda:

```python
# Exemplo de programação funcional
numeros = [1, 2, 3, 4, 5]

# Usando a função map com lambda
quadrados = list(map(lambda x: x ** 2, numeros))
print(f"Os quadrados dos números são: {quadrados}")
```

# Orientação a objetos
Python permite criar classes e objetos para modelar abstrações do mundo real.

```python
# Exemplo de orientação a objetos
class Pessoa:
    def __init__(self, nome, idade): # Esse é o construtor
        self.nome = nome
        self.idade = idade

    def cumprimentar(self):
        return f"Olá, meu nome é {self.nome} e eu tenho {self.idade} anos."

# Criando um objeto
pessoa = Pessoa("Ana", 25)
print(pessoa.cumprimentar())
```

#Listas

Listas permitem armazenar coleções ordenadas de elementos de diferentes tipos, e suas funcionalidades tornam-nas adequadas para diversas tarefas de programação.

## Características das listas

1. **Mutáveis:** Você pode modificar uma lista após criá-la (adicionar, remover ou alterar elementos).

2. **Dinâmicas:** Não têm tamanho fixo; podem crescer ou encolher conforme necessário.
3. **Tipos mistos:** Uma lista pode conter elementos de tipos diferentes (embora, geralmente, seja preferível manter uniformidade).
4. **Indexadas:** Os elementos são acessados usando índices baseados em zero (o primeiro elemento tem índice 0).

##Criando listas
Você pode criar uma lista usando colchetes ([]) ou a função list():

```python
# Lista vazia
lista_vazia = []

# Lista com elementos
frutas = ["maçã", "banana", "laranja"]

# Lista com tipos variados
misturada = [42, "texto", 3.14, True]
```

## Principais operações com listas


```python
# 1. Acessar elementos
frutas = ["maçã", "banana", "laranja"]
print(frutas[0])  # Saída: maçã
print(frutas[-1])  # Saída: laranja (índice negativo acessa de trás para frente)
```

```python
# 2. Modificar elementos
frutas[1] = "morango"
print(frutas)  # Saída: ['maçã', 'morango', 'laranja']
```

```python
# 3. Adcionar elementos
frutas.append("uva")  # Adiciona ao final
frutas.insert(1, "pera")  # Insere na posição 1
print(frutas)  # Saída: ['maçã', 'pera', 'morango', 'laranja', 'uva']
```

```python
# 4. Remover elementos
frutas.remove("morango")  # Remove o primeiro elemento correspondente
print(frutas)  # Saída: ['maçã', 'pera', 'laranja', 'uva']
del frutas[2]  # Remove pelo índice
print(frutas)  # Saída: ['maçã', 'pera', 'uva']
```

```python
# 5. Iterar sobre uma lista
for fruta in frutas:
    print(fruta)
```

```python
# 6. Ordem
numeros = [3, 1, 4, 1, 5, 9]
numeros.sort()  # Ordena em ordem crescente
print(numeros)  # Saída: [1, 1, 3, 4, 5, 9]

numeros.reverse()  # Inverte a ordem
print(numeros)  # Saída: [9, 5, 4, 3, 1, 1]
```

```python
# 7. List Comprehension

# Criar uma lista com os quadrados dos números de 0 a 9
quadrados = [x**2 for x in range(10)]
print(quadrados)  # Saída: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
```

```python
# 8. Listas aninhadas

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Acessar elementos de uma lista aninhada
print(matriz[1][2])  # Saída: 6 (linha 2, coluna 3)
```

# Arrays - vetores

Diferentemente de linguagens como C ou Java, Python não possui um tipo de dado array nativo. No entanto, oferece diversas alternativas que atendem a diferentes necessidades de programação.

O módulo `array` da biblioteca padrão permite criar arrays compactos e com tipos restritos, semelhantes aos encontrados em linguagens de tipagem estática.

Para computação numérica e científica, a biblioteca NumPy é o padrão mais amplamente usado. Seu tipo `ndarray` (matriz n-dimensional) oferece funcionalidade avançada, superando a estrutura básica de array do módulo padrão.

```python
# 1. Criação de um numpy array
import numpy as np
my_array = np.array([0, 1, 2, 3, 4])
print(my_array)
```

```python
#2. Operações

import numpy as np

array_A = np.array([1, 2, 3])
array_B = np.array([4, 5, 6])

print(array_A + array_B)
```

O NumPy é importante porque suas operações são significativamente mais rápidas do que os loops em Python, especialmente para matrizes grandes, devido à sua implementação otimizada em C.
No exemplo a seguir, comparamos a multiplicação e a adição de duas matrizes, `A` e `B`, usando ambas as abordagens.

```python
import numpy as np
import time

# Tamanhos da matriz
n, m = 100, 100

# Criando matrizes de exemplo
A = np.random.rand(n, m)
B = np.random.rand(n, m)

### Usando loops for em Python
start_time = time.time()

# Produto elemento a elemento com for
produto_for = [[A[i][j] * B[i][j] for j in range(m)] for i in range(n)]

# Soma das linhas com for
soma_linhas_for = [sum(produto_for[i]) for i in range(n)]

time_for = time.time() - start_time
print(f"Tempo com for: {time_for:.6f} segundos")

### Usando NumPy
start_time = time.time()

# Produto elemento a elemento com NumPy
produto_numpy = A * B

# Soma das linhas com NumPy
soma_linhas_numpy = produto_numpy.sum(axis=1)

time_numpy = time.time() - start_time
print(f"Tempo com NumPy: {time_numpy:.6f} segundos")

### Comparando os resultados
print("\nResultados são iguais:")
print("Produto elemento a elemento:", np.allclose(produto_for, produto_numpy))
print("Soma das linhas:", np.allclose(soma_linhas_for, soma_linhas_numpy))
```

##Criando um Gráfico de Linha


```python
import matplotlib.pyplot as plt

# Dados
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Criando o gráfico
plt.plot(x, y, label="Crescimento")  # Adiciona uma legenda
plt.xlabel("X - Eixo")  # Rótulo do eixo X
plt.ylabel("Y - Eixo")  # Rótulo do eixo Y
plt.title("Gráfico de Linha")  # Título do gráfico
plt.legend()  # Exibe a legenda
plt.grid(True)  # Adiciona uma grade
plt.show()  # Exibe o gráfico
```

## Gráfico de Barras


```python
import matplotlib.pyplot as plt

# Dados
categorias = ["A", "B", "C", "D"]
valores = [10, 15, 7, 12]

# Criando o gráfico
plt.bar(categorias, valores, color="blue", alpha=0.7)
plt.title("Gráfico de Barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

## Gráfico de Dispersão


```python
import matplotlib.pyplot as plt

# Dados
x = [5, 7, 8, 7, 2, 17, 2, 9]
y = [99, 86, 87, 88, 100, 86, 103, 87]

# Criando o gráfico
plt.scatter(x, y, color="red")
plt.title("Gráfico de Dispersão")
plt.xlabel("Idade")
plt.ylabel("Pontuação")
plt.show()
```

## Histogramas

```python
import matplotlib.pyplot as plt

# Dados
valores = [7, 8, 5, 4, 6, 9, 10, 6, 7, 6, 5, 4, 7]

# Criando o histograma
plt.hist(valores, bins=5, color="green", edgecolor="black")
plt.title("Histograma")
plt.xlabel("Intervalos")
plt.ylabel("Frequência")
plt.show()
```
